# Observations

We have collected loss observations for the 2024 Porto Alegre floods from different sources. These will be used to calibrate the model.

The observations contain different measures of impact: both affected and financial impact.

```python
from pathlib import Path

# -----------------------------------------------
# Set up paths to data
# -----------------------------------------------

OBSERVATIONS_DIR = "../data/observations/"
OBSERVATIONS_FILE = "observations_2024.csv"
observations_filepath = Path(OBSERVATIONS_DIR) / OBSERVATIONS_FILE
```

```python
import pandas as pd
import numpy as np
from IPython.display import display

# -----------------------------------------------
# Read and print the observations
# -----------------------------------------------

obs = pd.read_csv(observations_filepath)

print("Observations: affected")
obs_affected = obs[obs['Impact type'] == 'Affected']
display(obs_affected)

print("Observations: economic loss")
obs_loss = obs[obs['Impact type'] == 'Economic loss']
display(obs_loss)
```
